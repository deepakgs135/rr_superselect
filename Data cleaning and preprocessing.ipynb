```python
# Comprehensive batting dataset: delivery fact -> player-innings fact -> profile/context tables
# Uses ball_df_clean as the one-row-per-delivery source. Fielding data is not used.
import numpy as np
import pandas as pd

# Preserve source row count and inspect required schemas.
_bat_source = ball_df_clean.copy()
_bat_source_rows = len(_bat_source)
_bat_required = {'match_id','innings','striker_id','striker_name','batting_team_id','bowling_team_id','runs'}
_bat_missing_cols = _bat_required - set(_bat_source.columns)
if _bat_missing_cols:
    raise KeyError(f'Missing batting source columns: {_bat_missing_cols}')

# Match and competition dimensions are joined only after checking their key grain.
_match_cols = [c for c in ['match_id','comp_id','match_number','match_date','stage','venue_id','venue_country','home_team_id','home_team','away_team_id','away_team','home_team_captain_id','away_team_captain_id','toss_win_team_id','toss_opted','man_of_the_match','first_inn_revised_overs','second_inn_revised_overs','revised_target','knock_out','win_team_id','result'] if c in matches.columns]
_match_dim = matches[_match_cols].drop_duplicates()
if _match_dim['match_id'].duplicated().any():
    raise ValueError('matches has conflicting rows per match_id; resolve before context join')
_comp_cols = [c for c in ['comp_id','comp_name','match_format','season'] if c in competitions.columns]
_comp_dim = competitions[_comp_cols].drop_duplicates()
if 'comp_id' in _comp_dim and _comp_dim['comp_id'].duplicated().any():
    raise ValueError('competitions has conflicting rows per comp_id')

# Attach unique match and competition metadata at delivery grain; validate row preservation.
_bat_deliveries = _bat_source.merge(_match_dim, on='match_id', how='left', validate='many_to_one', suffixes=('','_match'))
if 'comp_id' in _bat_deliveries and 'comp_id' in _comp_dim:
    _bat_deliveries = _bat_deliveries.merge(_comp_dim, on='comp_id', how='left', validate='many_to_one', suffixes=('','_competition'))
if len(_bat_deliveries) != _bat_source_rows:
    raise AssertionError('Match/competition enrichment changed delivery row count')

# Team labels are competition-specific in this data. Use a validated (comp_id, team_id) dimension.
if {'comp_id','team_id','team_name'}.issubset(teams.columns) and 'comp_id' in _bat_deliveries:
    _team_dim = teams[['comp_id','team_id','team_name']].drop_duplicates()
    if _team_dim.duplicated(['comp_id','team_id']).any():
        raise ValueError('teams has conflicting rows per (comp_id, team_id)')
    _bat_deliveries = _bat_deliveries.merge(_team_dim.rename(columns={'team_id':'batting_team_id','team_name':'batting_team_name'}), on=['comp_id','batting_team_id'], how='left', validate='many_to_one')
    _bat_deliveries = _bat_deliveries.merge(_team_dim.rename(columns={'team_id':'bowling_team_id','team_name':'opposition_team_name'}), on=['comp_id','bowling_team_id'], how='left', validate='many_to_one')
else:
    _bat_deliveries['batting_team_name'] = np.nan
    _bat_deliveries['opposition_team_name'] = np.nan

# Explicit batting measures. 'runs' is the striker's runs; extras columns are not used as batter runs.
_bat_deliveries['runs'] = pd.to_numeric(_bat_deliveries['runs'], errors='coerce').fillna(0)
_bat_deliveries['balls_faced_flag'] = ~_bat_deliveries.get('extras_type', pd.Series(index=_bat_deliveries.index, dtype='object')).astype('string').str.lower().eq('wide').fillna(False)
_bat_deliveries['four_flag'] = _bat_deliveries['runs'].eq(4)
_bat_deliveries['six_flag'] = _bat_deliveries['runs'].eq(6)
_bat_deliveries['dismissed_striker_flag'] = _bat_deliveries.get('dismissed_player_id', pd.Series(index=_bat_deliveries.index, dtype='float')).eq(_bat_deliveries['striker_id'])
# The single Retire Out event has no dismissed_player_id; count it against striker by delivery event.
if 'dismissal_type' in _bat_deliveries:
    _retire_out = _bat_deliveries['dismissal_type'].astype('string').str.casefold().eq('retire out').fillna(False)
    _bat_deliveries['dismissed_striker_flag'] |= _retire_out & _bat_deliveries['is_wicket'].fillna(False).astype(bool) if 'is_wicket' in _bat_deliveries else _retire_out

# Batter innings is the base summary grain; no bowling aggregate is joined into batting facts.
_bat_keys = ['match_id','innings','striker_id','striker_name','batting_team_id']
_bat_group = _bat_deliveries.groupby(_bat_keys, dropna=False, sort=False)
player_batting_innings = _bat_group.agg(
    runs=('runs','sum'), balls_faced=('balls_faced_flag','sum'), fours=('four_flag','sum'), sixes=('six_flag','sum'),
    dismissed=('dismissed_striker_flag','max'), first_ball=('innings_ball_num','min') if 'innings_ball_num' in _bat_deliveries else ('runs','size'),
    batting_order=('batting_order','first') if 'batting_order' in _bat_deliveries else ('runs','size')
).reset_index()
player_batting_innings['dismissed'] = player_batting_innings['dismissed'].astype(bool)
player_batting_innings['not_out'] = ~player_batting_innings['dismissed']
player_batting_innings['strike_rate'] = np.where(player_batting_innings['balls_faced'].gt(0), 100*player_batting_innings['runs']/player_batting_innings['balls_faced'], np.nan)
player_batting_innings['boundary_runs'] = 4*player_batting_innings['fours'] + 6*player_batting_innings['sixes']
player_batting_innings['boundary_run_pct'] = np.where(player_batting_innings['runs'].gt(0), 100*player_batting_innings['boundary_runs']/player_batting_innings['runs'], np.nan)
player_batting_innings['is_50'] = player_batting_innings['runs'].between(50,99)
player_batting_innings['is_100'] = player_batting_innings['runs'].ge(100)
player_batting_innings['is_duck'] = player_batting_innings['runs'].eq(0)
player_batting_innings['high_score_not_out'] = player_batting_innings['not_out'].map({True:'*',False:''})

# Attach one match-context record to each player-innings row.
_player_context_cols = [c for c in _match_cols if c != 'match_id'] + [c for c in ['comp_name','match_format','season'] if c in _bat_deliveries.columns]
_player_match_context = _bat_deliveries[['match_id'] + _player_context_cols].drop_duplicates('match_id')
player_batting_innings = player_batting_innings.merge(_player_match_context, on='match_id', how='left', validate='many_to_one')
if 'match_date' in player_batting_innings:
    player_batting_innings['match_date'] = pd.to_datetime(player_batting_innings['match_date'], errors='coerce')
    player_batting_innings['year'] = player_batting_innings['match_date'].dt.year
if {'home_team_id','away_team_id'}.issubset(player_batting_innings.columns):
    player_batting_innings['batting_home_away'] = np.select([player_batting_innings['batting_team_id'].eq(player_batting_innings['home_team_id']),player_batting_innings['batting_team_id'].eq(player_batting_innings['away_team_id'])],['Home','Away'],default='Neutral/Unknown')
    player_batting_innings['opposition_team_id'] = np.where(player_batting_innings['batting_team_id'].eq(player_batting_innings['home_team_id']),player_batting_innings['away_team_id'],player_batting_innings['home_team_id'])
if 'toss_win_team_id' in player_batting_innings:
    player_batting_innings['batting_toss_winner'] = player_batting_innings['batting_team_id'].eq(player_batting_innings['toss_win_team_id'])
if 'win_team_id' in player_batting_innings:
    player_batting_innings['batting_team_won'] = player_batting_innings['batting_team_id'].eq(player_batting_innings['win_team_id'])

# Overall player profiles: means use innings aggregates; batting average excludes not-outs.
_bat_profile = player_batting_innings.groupby(['striker_id','striker_name'], dropna=False, sort=False)
player_batting_profile = _bat_profile.agg(
    runs=('runs','sum'), innings=('match_id','size'), not_outs=('not_out','sum'), dismissals=('dismissed','sum'), balls_faced=('balls_faced','sum'),
    fours=('fours','sum'), sixes=('sixes','sum'), highest_score=('runs','max'), fifties=('is_50','sum'), centuries=('is_100','sum'), ducks=('is_duck','sum')
).reset_index().rename(columns={'striker_id':'player_id','striker_name':'player_name'})
player_batting_profile['average'] = np.where(player_batting_profile['dismissals'].gt(0), player_batting_profile['runs']/player_batting_profile['dismissals'], np.nan)
player_batting_profile['strike_rate'] = np.where(player_batting_profile['balls_faced'].gt(0), 100*player_batting_profile['runs']/player_batting_profile['balls_faced'], np.nan)
player_batting_profile['boundary_runs'] = 4*player_batting_profile['fours'] + 6*player_batting_profile['sixes']
player_batting_profile['boundary_run_pct'] = np.where(player_batting_profile['runs'].gt(0),100*player_batting_profile['boundary_runs']/player_batting_profile['runs'],np.nan)
player_batting_profile['balls_per_four'] = np.where(player_batting_profile['fours'].gt(0),player_batting_profile['balls_faced']/player_batting_profile['fours'],np.nan)
player_batting_profile['balls_per_six'] = np.where(player_batting_profile['sixes'].gt(0),player_batting_profile['balls_faced']/player_batting_profile['sixes'],np.nan)
player_batting_profile['balls_per_50_plus_innings'] = np.where((player_batting_profile['fifties']+player_batting_profile['centuries']).gt(0),player_batting_profile['balls_faced']/(player_batting_profile['fifties']+player_batting_profile['centuries']),np.nan)
player_batting_profile['balls_per_century_innings'] = np.where(player_batting_profile['centuries'].gt(0),player_batting_profile['balls_faced']/player_batting_profile['centuries'],np.nan)
for _n in [0,1,2,3,4,6]:
    _bat_profile_runs = _bat_deliveries.assign(_run_bin=_bat_deliveries['runs'].where(_bat_deliveries['runs'].isin([0,1,2,3,4,6]),-1)).groupby(['striker_id','striker_name'])['_run_bin'].apply(lambda s: s.eq(_n).sum()).rename(f'scoring_{_n}_balls').reset_index()
    player_batting_profile = player_batting_profile.merge(_bat_profile_runs, left_on=['player_id','player_name'], right_on=['striker_id','striker_name'], how='left', validate='one_to_one').drop(columns=['striker_id','striker_name'])

# Dismissal-type counts are keyed to actual dismissed player; excludes unrelated/non-striker wicket events.
if 'dismissal_type' in _bat_deliveries:
    batting_dismissal_types = (_bat_deliveries.loc[_bat_deliveries['dismissed_striker_flag']]
        .groupby(['striker_id','striker_name','dismissal_type'],dropna=False).size().rename('dismissal_count').reset_index()
        .rename(columns={'striker_id':'player_id','striker_name':'player_name'}))
else:
    batting_dismissal_types = pd.DataFrame(columns=['player_id','player_name','dismissal_type','dismissal_count'])

# Detailed context outputs share the same innings fact, including match date/year, venue, opposition, team, comp/season/format, stage, toss and result fields.
_context_dims = [c for c in ['venue_id','venue_country','opposition_team_id','opposition_team_name','batting_team_id','batting_team_name','comp_id','comp_name','season','match_format','stage','batting_home_away','toss_win_team_id','toss_opted','batting_toss_winner','win_team_id','result','batting_team_won','match_date','year'] if c in player_batting_innings.columns]
def _make_bat_context(dims):
    keys=['striker_id','striker_name']+dims
    g=player_batting_innings.groupby(keys,dropna=False,sort=False)
    out=g.agg(runs=('runs','sum'),innings=('match_id','size'),not_outs=('not_out','sum'),dismissals=('dismissed','sum'),balls_faced=('balls_faced','sum'),fours=('fours','sum'),sixes=('sixes','sum'),fifties=('is_50','sum'),centuries=('is_100','sum'),ducks=('is_duck','sum')).reset_index()
    out['average']=np.where(out['dismissals'].gt(0),out['runs']/out['dismissals'],np.nan)
    out['strike_rate']=np.where(out['balls_faced'].gt(0),100*out['runs']/out['balls_faced'],np.nan)
    out['boundary_runs']=4*out['fours']+6*out['sixes']
    out['boundary_run_pct']=np.where(out['runs'].gt(0),100*out['boundary_runs']/out['runs'],np.nan)
    return out
batting_context_profile = _make_bat_context(_context_dims)
batting_venue_profile = _make_bat_context([c for c in ['venue_id','venue_country'] if c in player_batting_innings.columns])
batting_opposition_profile = _make_bat_context([c for c in ['opposition_team_id','opposition_team_name'] if c in player_batting_innings.columns])
batting_comp_season_profile = _make_bat_context([c for c in ['comp_id','comp_name','season','match_format'] if c in player_batting_innings.columns])
batting_team_profile = _make_bat_context([c for c in ['batting_team_id','batting_team_name'] if c in player_batting_innings.columns])
batting_match_context = player_batting_innings.copy()

# QA summary and invariants.
batting_qa = pd.Series({
    'source_delivery_rows': _bat_source_rows,
    'enriched_delivery_rows': len(_bat_deliveries),
    'player_innings_rows': len(player_batting_innings),
    'unique_batters': player_batting_profile['player_id'].nunique(),
    'source_missing_striker_or_innings': int(_bat_source[['striker_id','innings']].isna().any(axis=1).sum()),
    'runs_source_sum': float(_bat_source['runs'].sum()),
    'runs_player_innings_sum': float(player_batting_innings['runs'].sum()),
    'wide_deliveries_excluded_from_balls_faced': int(_bat_deliveries['extras_type'].astype('string').str.lower().eq('wide').sum()) if 'extras_type' in _bat_deliveries else 0,
    'wicket_events': int(_bat_deliveries['is_wicket'].fillna(False).astype(bool).sum()) if 'is_wicket' in _bat_deliveries else 0,
    'dismissed_batter_innings': int(player_batting_innings['dismissed'].sum()),
    'unique_player_innings_keys': int(player_batting_innings[['match_id','innings','striker_id','batting_team_id']].drop_duplicates().shape[0]),
})
if len(_bat_deliveries) != _bat_source_rows or not np.isclose(player_batting_innings['runs'].sum(), _bat_source['runs'].sum()):
    raise AssertionError('Batting QA invariant failed: row preservation or batter-run reconciliation')
print('Created: player_batting_innings, player_batting_profile, batting_dismissal_types, batting_context_profile, batting_venue_profile, batting_opposition_profile, batting_comp_season_profile, batting_team_profile, batting_match_context')
print(batting_qa.to_string())
display(player_batting_profile.sort_values(['runs','innings'],ascending=False).head(10))
```

```python
# Pressure batter summary under the 4-part definition:
# Innings = 2, Legal Balls Remaining <= 30, Wickets Down >= 3, Required Run Rate >= 15
_bbb = ball_df_clean[ball_df_clean['innings'].isin([1, 2])].copy()
for _c in ['overid', 'over_ball_num', 'innings_ball_num', 'runs', 'extras1_runs', 'extras_2_runs', 'is_wicket']:
    _bbb[_c] = pd.to_numeric(_bbb[_c], errors='coerce').fillna(0)
_ext = _bbb['extras_type'].astype(str).str.strip().str.lower()
_is_extra_1 = _ext.isin(['wide', 'overstep', 'byes', 'legbyes', 'other'])
_is_penalty = _ext.eq('penalty')
_bbb['team_runs_ball'] = _bbb['runs'] + _is_extra_1.astype(int) + 5 * _is_penalty.astype(int) + _bbb['extras_2_runs']
_bbb = _bbb.sort_values(['match_id', 'innings', 'innings_ball_num', 'overid', 'over_ball_num']).reset_index(drop=True)

_inn1 = _bbb[_bbb['innings'] == 1].groupby('match_id', as_index=False)['team_runs_ball'].sum().rename(columns={'team_runs_ball': 'inn1_runs'})
_m_target = matches[['match_id', 'second_inn_revised_overs', 'revised_target']].drop_duplicates('match_id').merge(_inn1, on='match_id', how='left')
_m_target['target'] = np.where(_m_target['revised_target'].notna() & (_m_target['revised_target'] > 0), _m_target['revised_target'], _m_target['inn1_runs'] + 1)
_rev = pd.to_numeric(_m_target['second_inn_revised_overs'], errors='coerce')
_m_target['max_balls'] = np.where(_rev.notna() & (_rev > 0), _rev, 120.0)

_bbb = _bbb.merge(_m_target[['match_id', 'target', 'max_balls']], on='match_id', how='left')
_bbb['wickets_down'] = _bbb.groupby(['match_id', 'innings'])['is_wicket'].cumsum() - _bbb['is_wicket']
_bbb['team_runs_before'] = _bbb.groupby(['match_id', 'innings'])['team_runs_ball'].cumsum() - _bbb['team_runs_ball']

_is_wide = _bbb['extras_type'].astype(str).str.lower().eq('wide')
_is_noball = _bbb['extras_type'].astype(str).str.lower().eq('overstep')
_bbb['is_legal'] = (~_is_wide & ~_is_noball).astype(int)
_bbb['legal_before'] = _bbb.groupby(['match_id', 'innings'])['is_legal'].cumsum() - _bbb['is_legal']
_bbb['balls_left'] = _bbb['max_balls'] - _bbb['legal_before']
_bbb['runs_needed'] = _bbb['target'] - _bbb['team_runs_before']
_bbb['rrr'] = np.where(_bbb['balls_left'] > 0, 6.0 * _bbb['runs_needed'] / _bbb['balls_left'], np.nan)

_faced = _bbb[~_is_wide].copy()
_faced['is_pressure'] = (
    (_faced['innings'] == 2)
    & (_faced['balls_left'] > 0)
    & (_faced['balls_left'] <= 30)
    & (_faced['wickets_down'] >= 3)
    & (_faced['rrr'] >= 15.0)
)
_p_grp = _faced[_faced['is_pressure']].groupby(['striker_id', 'striker_name'], as_index=False).agg(
    pressure_balls=('runs', 'size'), pressure_runs=('runs', 'sum')
)
_o_grp = _faced[~_faced['is_pressure']].groupby(['striker_id', 'striker_name'], as_index=False).agg(
    ordinary_balls=('runs', 'size'), ordinary_runs=('runs', 'sum')
)
pressure_batter_summary = _p_grp.merge(_o_grp, on=['striker_id', 'striker_name'], how='inner')
pressure_batter_summary['pressure_sr'] = 100.0 * pressure_batter_summary['pressure_runs'] / pressure_batter_summary['pressure_balls']
pressure_batter_summary['ordinary_sr'] = 100.0 * pressure_batter_summary['ordinary_runs'] / pressure_batter_summary['ordinary_balls']
pressure_batter_summary = pressure_batter_summary.sort_values('pressure_balls', ascending=False)
pressure_batter_summary.to_csv(r'D:\RR\cleaned_data\pressure_batter_summary_rrr15.csv', index=False)
display(pressure_batter_summary.head(10))
```
